```python
import pandas as pd
import statsmodels.api as sm
```

```python
students = pd.DataFrame({
    "Hours":[1,2,3,4,5],
    "Marks":[45,55,65,75,85]
})
```

```python
X = students["Hours"]
y = students["Marks"]
```

```python
X = sm.add_constant(X)
```

```python
model = sm.OLS(y, X)
```

```python
results = model.fit()
```

```python
print(results.summary())
```
